## 2. Selected Dataset & Justification

**Selected: Irrigation Scheduling Dataset** (`GCEKIoTCommunity/Irrigation-Dataset`).

- Its target column, **`Irrigation(Y/N)`**, is exactly the decision the app needs to make:
  "does this plant need watering right now, given current conditions?"
- Its feature columns — **Soil Moisture, Soil Temperature, Temperature, Humidity**, plus
  `CropType` and `CropDays` — line up directly with the sensor/weather inputs the app plans
  to use, with humidity front and center as originally described.
- It is small and clean enough to explore fully in this notebook, and openly hosted (no
  API key or paywall), which made it possible to verify every value below rather than just
  read a description.

**Limitation:** only 150 readings across two crop types — good for this exploration exercise,
but the production model will likely need a larger sensor/weather dataset behind it.

```python
import pandas as pd
import numpy as np

# Publicly hosted dataset
URL = "https://raw.githubusercontent.com/GCEKIoTCommunity/Irrigation-Dataset/master/Project_datasheet_2019-2020.xlsx"

cols = ["CropType", "CropDays", "Soil Moisture", "Soil Temperature", "Temperature", "Humidity", "Irrigation(Y/N)"]
df = pd.read_excel(URL, usecols=cols)
df.head()
```

### a. How many rows and columns are contained in the dataset? (3 marks)

```python
n_rows, n_cols = df.shape
print(f"Rows: {n_rows}")
print(f"Columns: {n_cols}")
df.shape
```

### b. What datatypes are contained in the dataset? (3 marks)

### c. Is the dataset complete? i.e. no missing values? (2 marks)

```python
missing_per_column = df.isnull().sum()
print(missing_per_column)
print()
total_missing = df.isnull().sum().sum()
print(f"Total missing values: {total_missing}")
print("Dataset is complete (no missing values)." if total_missing == 0 else "Dataset has missing values.")
```

```python
first_15 = df.head(15)
last_20 = df.tail(20)

df_sample = pd.concat([first_15, last_20]).reset_index(drop=True)
print(f"df_sample shape: {df_sample.shape}")
df_sample
```
